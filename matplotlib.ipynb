# matplotlib 学习簿

`matplotlib` 是 `Python` 中最常用的绘图库之一，广泛用于数据可视化。它提供了丰富的绘图功能，可生成折线图、柱状图、散点图、饼图、直方图等多种图形。

`pyplot` 是 `matplotlib` 的核心模块，提供类似 MATLAB 的绘图接口。它以函数形式控制图形的创建、显示和样式设置，支持绘制折线图、柱状图、散点图等多种图表。通过简单命令即可快速生成高质量可视化结果，广泛应用于数据分析与科研绘图。通常来说，我们使用简短的别名 `plt` 来导入它，`%matplotlib inline` 设定让图形在 Notebook 中直接显示出来。

````python
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
````

## 常见函数

| 函数名                 | 功能说明                |
| :------------------ | :------------------ |
| `plot()`            | 绘制折线图或点线图，是最常用的绘图函数 |
| `bar()`             | 绘制垂直柱状图             |
| `barh()`            | 绘制水平柱状图             |
| `scatter()`         | 绘制散点图，用于显示变量间关系     |
| `pie()`             | 绘制饼图，展示各部分占比        |
| `hist()`            | 绘制直方图，用于展示数据分布      |
| `boxplot()`         | 绘制箱线图，显示数据分布的离散程度   |
| `imshow()`          | 显示图像数据（如矩阵或图片）      |
| `title()`           | 设置图表标题              |
| `xlabel()`          | 设置 `x` 轴标签            |
| `ylabel()`          | 设置 `y` 轴标签            |
| `legend()`          | 添加图例                |
| `xlim()` / `ylim()` | 设置坐标轴显示范围           |
| `grid()`            | 显示或隐藏网格线            |
| `show()`            | 显示绘制的图形             |

## 基本操作

### 绘制点和线

````python
x1 = [0, 1, 2, 3, 4]
y1 = [0, 2, 6, 2, 8]

plt.plot(x1, y1, color='blue', linestyle='--', linewidth=2, marker='s')
plt.title('Line Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
````

在上图中，我们利用到了 `plot()` 函数，`plot()` 函数是绘制二维图形的基础，我们还可以利用 `scatter()` 函数来绘制点，如下图所示：

````python
x2 = [1, 2, 3, 4, 5]
y2 = [2, 9, 3, 7, 11]

plt.scatter(x2, y2, color='red', marker='o', label='Data Points')
plt.title('Scatter Plot Example')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.grid(True)
plt.show()
````

从上面的两张图里也能看出，图上面的点、线是可以通过设置属性变得更加**美观**的，我们来列一个表格简单介绍一下所有的这些属性应该怎么使用：

| 属性                              | 参数类型 / 示例                                              | 说明                           |
| :------------------------------ | :----------------------------------------------------- | :--------------------------- |
| `color` / `c`                   | `'red'`, `'blue'`, `'g'`, `'#FF8800'`                  | 设置颜色，可用颜色名、简写或十六进制值          |
| `linestyle` / `ls`              | `'-'`, `'--'`, `':'`, `'-.'`, `'none'`                 | 设置线型（实线、虚线、点线、点划线、无线）        |
| `linewidth` / `lw`              | `1`, `2.5`, `5`                                        | 设置线宽                         |
| `marker`                        | `'o'`, `'s'`, `'^'`, `'x'`, `'+'`, `'d'`, `'*'`, `'.'` | 设置点的形状                       |
| `markersize` / `ms`             | `6`, `10`, `15`                                        | 设置点的大小                       |
| `markeredgecolor` / `mec`       | `'black'`, `'k'`, `'#333333'`                          | 设置点边缘颜色                      |
| `markerfacecolor` / `mfc`       | `'white'`, `'red'`, `'none'`                           | 设置点内部填充颜色                    |
| `alpha`                       | `0.0 ~ 1.0`                                            | 设置透明度，数值越小越透明                |
| `label`                       | `'Data1'`, `'sin(x)'`                                  | 设置图例文字，与 `plt.legend()` 配合使用 |
| `grid()`                      | `plt.grid(True)`                                       | 是否显示网格线                      |
| `xlim(), ylim()`              | `plt.xlim(0,10)`                                       | 设置坐标轴显示范围                    |
| `title(), xlabel(), ylabel()` | 文本字符串                                                  | 设置标题和坐标轴标签                   |

下面的图形是使用了诸多属性后的示例，你也可以在此基础上继续进行美观调整：

````python
x3 = np.linspace(0, 3, 7)
y3 = 2 ** x3 - 5

plt.plot(
    x3, y3,
    color='purple',             # 线的颜色
    linestyle='--',             # 虚线
    linewidth=2,                # 线宽
    marker='o',                 # 圆形标记
    markersize=8,               # 标记大小
    markerfacecolor='white',    # 标记填充色
    markeredgecolor='black',    # 标记边框色
    alpha=0.8,                  # 透明度
    label='y = 2x^2-5'
)

plt.title('Line and Marker Style Example')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
````

### 绘制函数图

在 `matplotlib` 中，我们可以方便地绘制**函数**曲线：

````python
x4 = np.linspace(0, 2 * np.pi, 200)  # 在 0 到 2π 之间取 200 个点
y4 = np.sin(x4)

plt.plot(x4, y4, label='y = sinx')
plt.title('sinx')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
````

在同一张图里，也可以放两条或更多函数的曲线：

````python
x5 = np.linspace(0, 2 * np.pi, 200)  # 在 0 到 2π 之间取 200 个点
y5 = np.sin(x5)
y6 = np.cos(x5)

plt.plot(x5, y5, label='sin')
plt.plot(x5, y6, linestyle='--', label='cos')
plt.xlabel('x')
plt.ylabel('y')
plt.title('sin & cos')
plt.legend()
plt.show()
````

### 绘制柱形图

* `bar(x, height)`：绘制垂直柱形图
* `color`：柱子颜色
* `width`：柱宽
* 若需水平展示，可使用 `plt.barh()`

````python
categories = ['A', 'B', 'C', 'D']
values = [25, 40, 30, 10]

plt.bar(categories, values, color='orange', width=0.6)
plt.title('Bar Chart Example')
plt.xlabel('Category')
plt.ylabel('Value')
plt.show()
````

### 绘制饼状图

* `labels`：每部分标签
* `sizes`：每部分占比
* `autopct`：显示百分比格式
* `explode`：突出某部分
* `startangle`：旋转起始角度

````python
labels = ['Apple', 'Banana', 'Cherry', 'Durian']
sizes = [30, 25, 25, 20]
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
explode = [0.05, 0, 0, 0]  # 突出第一个部分

plt.pie(
    sizes,
    labels=labels,
    colors=colors,
    explode=explode,
    autopct='%1.1f%%',
    shadow=True,
    startangle=90
)
plt.title('Pie Chart Example')
plt.show()
````

### 绘制直方图

* `hist(data)`：绘制直方图
* `bins`：直方图的区间数
* `edgecolor`：柱子边框颜色
* `alpha`：透明度
* 常用于连续型数据的分布可视化

````python
import numpy as np

data = np.random.randn(1000)  # 生成1000个符合正态分布的随机数

plt.hist(data, bins=20, color='skyblue', edgecolor='black', alpha=0.8)
plt.title('Histogram Example')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
````

### 操作图像

`matplotlib` 还可以对图像进行操作，包括从文件中读取图像和保存图像，也能对图像进行简单处理（如灰度化、翻转、裁剪等）。

下面的代码可以从文件中读取图像。

顺便提一下，我们使用的示例图像是非常著名的 *Lena*：

 > Lenna or Lena is the name given to a standard test image widely used in the field of image processing since 1973. It is a picture of the Swedish model Lena Söderberg (Lena Forsén), shot by photographer Dwight Hooker, cropped from the centerfold of the November 1972 issue of Playboy magazine. The spelling "Lenna" came from the model's desire to encourage the proper pronunciation of her name. "I didn't want to be called Leena," she explained.

````python
import matplotlib.image as mpimg

image = mpimg.imread('assets/lena.png')
print(image.shape)
plt.imshow(image)
````

上面显示读取出来的图像数据是一个三维数组，其中前两个维度是对应坐标像素点的数值（默认为 `0` 到 `1` 之间的一个浮点数）；第三个维度则表示数值对应的颜色通道，`0` 表示红色通道，`1` 表示绿色通道，`2` 表示蓝色通道。

#### 翻转

由于我们处理的图片本质上就是一个 `numpy` 数组，我们可以使用 `numpy` 提供的翻转函数来实现对图片的翻转，`np.fliplr()` 实现左右翻转，`np.flipud()` 实现上下翻转。

````python
h_flipped_image = np.fliplr(image)  # 水平翻转
plt.imshow(h_flipped_image)
````

````python
v_flipped_image = np.flipud(image)  # 垂直翻转
plt.imshow(v_flipped_image)
````

#### 旋转

`matplotlib` 本身不提供旋转图像的直接函数，但我们可以使用 `scipy.ndimage` 库中的 `rotate` 函数来实现图像的旋转。

下面的代码可以将图像旋转一定角度：

````python
from scipy import ndimage

rotated_image = ndimage.rotate(image, 45, reshape=False)
plt.imshow(rotated_image)
````

#### 裁剪

如前所述，我们处理的图片本质上就是一个 `numpy` 数组，我们可以通过数组的切片操作来实现对图片的裁剪。

````python
cropped_image = image[50:300, 50:300]
plt.imshow(cropped_image)
````

#### 明暗

通过调整图像数组的数值，我们可以调节图像的明暗程度。例如，将图像数组的值乘以一个小于 `1` 的系数即可使图像变暗；反之，乘以一个大于 `1` 的系数则会使图像变亮。

我们可以使用 `numpy` 的 `clip` 函数来确保图像数组的值仍然在有效范围内（通常是 `0` 到 `1` 之间或 `0` 到 `255` 之间，取决于数据类型）。

````python
darker_image = image.copy()

plt.figure(figsize=(10, 6))

for i in range(-2, 2):
    plt.subplot(2, 2, i+3)
    x = 1 + 0.2 * i
    plt.axis('off')
    plt.title('x={:.1f}'.format(x))
    darker_image = np.clip(darker_image, 0, 1)
    plt.imshow(darker_image * x)

plt.show()
````

#### RGB 颜色

`matplotlib` 中的图像通常以 RGB（三原色）格式存储和显示。每个像素由红、绿、蓝三个通道的强度值组成，范围通常在 `0` 到 `1` 之间（或 `0` 到 `255` 之间，取决于数据类型）。

我们可以通过调整这三个通道的值来改变图像的颜色。例如，将红色通道的值增加可以使图像变得更红，反之则可以使图像变得更冷。

以下是一个简单的示例，演示如何调整图像的 RGB 通道：

````python
color_adjusted_image = image.copy()
# lower red, increase blue
color_adjusted_image[:, :, 0] = np.clip(color_adjusted_image[:, :, 0] * 0.5, 0, 1)
color_adjusted_image[:, :, 2] = np.clip(color_adjusted_image[:, :, 2] * 1.5, 0, 1)

plt.imshow(color_adjusted_image)
plt.axis('off')
plt.show()
````

#### 灰度化

可以通过RGB三通道取平均来实现灰度化，`imshow` 函数中使用 `cmap='gray'` 参数可以更好地显示灰度图像。

````python
grayscale_image = image.mean(axis=2)
plt.imshow(grayscale_image, cmap='gray')
plt.axis('off')
plt.show()
````

### 绘制复合图

上面实际上已经用了一次，`matplotlib` 中可以使用 `subplot()` 绘制多个子图，将多个图整合在同一张画布中，我们称之为**复合图**。

方法 `plt.subplot(nrows, ncols, index)` 的参数有三个：
- `nrows`：子图的行数
- `ncols`：子图的列数
- `index`：当前绘制子图编号（注意，这个编号是从 `1` 开始的）

````python
x6 = np.arange(0, 3 * np.pi, 0.1)
y7 = np.sin(x6)
y8 = np.cos(x6)

# 两行一列布局，现在绘制第一个子图
plt.subplot(2, 1, 1)

plt.plot(x6, y7)
plt.title('sin')

# 布局不变，绘制第二个子图
plt.subplot(2, 1, 2)
plt.plot(x6, y8)
plt.title('cos')

plt.show()
````

## 进阶：理解 Axe

*Axe* 是 `matplotlib` 中的一个重要概念，代表图表中的一个绘图区。每个图表（Figure）可以包含多个 *Axe*，每个 *Axe* 可以包含自己的坐标轴、标题、标签和图形元素。

下面是 `matplotlib` 官方网站上的一张图，详细标出了 *Axe* 的各个组成部分：

![Anatomy of Matplotlib](assets/anatomy-of-matplotlib.png)

我们可以这么理解这些基本的概念：

* *Figure* 是一张画布；
* *Axes* 是画布上的一个子图区域；
* *Axis* 是隶属于区域的坐标轴。

从上图可以归纳出 `matplotlib` 的层次结构大致如下：

```
Figure（画布）
│
├── Axes（坐标系）
│    ├── Axis（坐标轴对象）
│    │     ├── XAxis
│    │     └── YAxis
│    ├── Title / Label / Legend 等文本对象
│    └── Artists（线条、柱状、点、图像等元素）
```

### 获取和操作 Axes

我们可以在绘图的过程中随时调用 `plt.gca()` 来获取当前绘制的 `Axe` 对象，从而对其进行更细粒度的控制和管理，`matplotlib` 提供的绝大部分操作都可以通过 `Axe` 对象来完成，例如设置坐标轴的刻度、标签、标题等。

> `gca` 代表 "get current Axes"。

````python
plt.figure()
ax = plt.gca()

ax.plot([1, 2, 3], [4, 5, 6])
ax.set_title("Linear")
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")

plt.show()
````

我们还可以调用 `plt.subplots()` 函数来创建一个包含多个 `Axe` 的图表，并返回 Figure 和 Axes 对象。这样我们就可以对每个 `Axe` 进行独立的操作和管理。下面这个例子演示了如何创建多个子图并通过 `Axe` 对象和 `plt` 管理子图和总图的各项属性：

````python
x = np.linspace(0, 2 * np.pi, 100)

fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2)

ax1.plot(x, np.sin(x))
ax1.set_title("Sine Wave")
ax2.plot(x, np.cos(x))
ax2.set_title("Cosine Wave")
ax3.plot(x, np.tan(x))
ax3.set_title("Tangent Wave")
ax4.plot(x, np.exp(x))
ax4.set_title("Exponential Function")

plt.suptitle("Multiple Subplots Example")
plt.tight_layout(rect=(0, 0.03, 1, 0.95))
plt.show()
````

小技巧：`subplots` 方法返回的 `axes` 是一个二维数组（当 `nrows` 和 `ncols` > 1 时），可以用 `axes.flat` 来将其变成一维数组进行循环访问。

### 自定义 Axes 布局

有时候我们会希望自己精确控制子图的位置和布局而非默认的网格布局，这时可使用 `plt.add_axes()` 方法，该方法允许我们通过指定坐标和尺寸来添加新的 `Axes` 对象。

````python
fig = plt.figure(figsize=(8, 6))
ax1 = fig.add_axes((0.1, 0.1, 0.8, 0.8))  # (left, bottom, width, height)
ax2 = fig.add_axes((0.2, 0.15, 0.25, 0.25))

x = np.linspace(0, 10, 100)
ax1.plot(x, np.sin(x))
ax2.plot(x, np.cos(x), 'r')
plt.show()
````

注意 `add_axes()` 方法所有参数都是相对于 *Figure* 的相对值（`0` 到 `1` 之间的浮点数）。
